```python
from openai import OpenAI

client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)

MODEL_NAME = "google/gemini-2.5-flash"

print("OpenRouter client configured successfully!")
```

```python
print("Please upload your PDF file:")

uploaded = files.upload()

pdf_filename = list(uploaded.keys())[0]

print("PDF uploaded successfully:")
print(pdf_filename)
```

```python
from pypdf import PdfReader

reader = PdfReader(pdf_filename)

pdf_text = ""

for page_number, page in enumerate(reader.pages):
    text = page.extract_text()

    if text:
        pdf_text += text + "\n"

print("PDF text extracted successfully!")
print("Number of characters:", len(pdf_text))
```

```python
print(pdf_text[:2000])
```

```python
def chunk_text(text, chunk_size=1000, overlap=200):
    chunks = []

    start = 0

    while start < len(text):
        end = start + chunk_size

        chunk = text[start:end]

        if chunk.strip():
            chunks.append(chunk.strip())

        start += chunk_size - overlap

    return chunks
```

```python
chunks = chunk_text(pdf_text)

print("Total PDF chunks:", len(chunks))

for i, chunk in enumerate(chunks[:3]):
    print("\n--- CHUNK", i + 1, "---")
    print(chunk[:500])
```

```python
embedder = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
embeddings = embedder.encode(chunks).tolist()

print("Embeddings created successfully!")
print("Number of embeddings:", len(embeddings))
```

```python
# Create ChromaDB client
chroma_client = chromadb.Client()

# Create collection
collection = chroma_client.get_or_create_collection(
    name="pdf_chatbot"
)

print("ChromaDB collection created successfully!")
```

```python
# Create unique IDs
ids = [f"chunk_{i}" for i in range(len(chunks))]

collection.add(
    documents=chunks,
    embeddings=embeddings,
    ids=ids
)

print("PDF chunks stored in ChromaDB successfully!")
print("Total documents in collection:", collection.count())
```

```python
def retrieve_pdf_context(query: str, top_k: int = 3) -> list[str]:

    # Convert user's question into embedding
    query_embedding = embedder.encode([query]).tolist()

    # Search ChromaDB
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=top_k
    )

    # Return retrieved PDF documents
    return results["documents"][0]
```

```python
def ask_pdf(query: str):

    # Retrieve relevant PDF passages
    context_passages = retrieve_pdf_context(query, top_k=3)

    # Combine retrieved passages
    context_str = "\n".join(
        f"- {p}" for p in context_passages
    )

    # Create prompt
    prompt = f"""
You are an intelligent document analysis assistant.

Answer the question using ONLY the information provided
in the PDF context below.

If the information is not contained within the provided
context, clearly state:

"I cannot find the answer in the provided PDF."

Do not make up information.

PDF Context:
{context_str}

Question:
{query}

Answer:
"""

    # Send request to OpenRouter
    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2
    )

    # Get answer
    answer = response.choices[0].message.content

    return answer, context_passages
```

```python
def ask_pdf(query: str):

    # Retrieve relevant PDF passages
    context_passages = retrieve_pdf_context(query, top_k=5)

    # Combine retrieved passages
    context_str = "\n\n".join(
        f"PASSAGE {i+1}:\n{p}"
        for i, p in enumerate(context_passages)
    )

    # Strong RAG prompt
    prompt = f"""
You are a PDF question-answering assistant.

You MUST answer the user's question using the PDF passages provided below.

IMPORTANT RULES:
1. Read the PDF passages carefully.
2. If the answer is explicitly present in the passages, answer it directly.
3. Do NOT say "I cannot find the answer" when the information is present.
4. Use only information from the provided PDF passages.
5. For questions asking about the main topic, give a short summary of what the PDF contains.
6. For factual questions, give a direct and concise answer.
7. Do not invent information.

PDF PASSAGES:
----------------
{context_str}
----------------

USER QUESTION:
{query}

ANSWER:
"""

    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0,
        max_tokens=500
    )

    answer = response.choices[0].message.content

    return answer, context_passages
```

```python
question = "What is Suraj Dilip Sanap's current degree?"

answer, context = ask_pdf(question)

print("========================================")
print("CHATBOT ANSWER")
print("========================================")
print(answer)
```
